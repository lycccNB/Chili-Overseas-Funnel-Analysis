```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from scipy.stats import kruskal

# 1. Load cleaned dataset
df_clean = pd.read_csv('chili_tiktok_clean_data.csv')

# 2. Aggregate median metrics by content category
summary = (
    df_clean.groupby('Category')
    .agg(
        N=('playCount', 'count'),
        Median_Views=('playCount', 'median'),
        Median_Like_Pct=('Like_Rate_Pct', 'median'),
        Median_Save_Pct=('Save_Rate_Pct', 'median'),
        Median_Intent_Pct=('Actionable_Intent_Pct', 'median'),
        Text_CTA_Pct=('Has_Text_CTA', lambda x: x.mean() * 100),
    )
    .reset_index()
    .sort_values(by='Median_Intent_Pct', ascending=False)
    .reset_index(drop=True)
)

# 3. Non-parametric significance test (Kruskal-Wallis)
groups_intent = [
    g['Actionable_Intent_Pct'].values for _, g in df_clean.groupby('Category')
]
h_stat, p_val = kruskal(*groups_intent)

print(
    f'TikTok China Travel Funnel Analysis (N = {len(df_clean)}, Kruskal-Wallis'
    f' H = {h_stat:.2f}, p = {p_val:.2e})'
)
print('-' * 88)
print(summary.round(2).to_string(index=False))
print('-' * 88)

# 4. Restrained monochrome chart (DejaVu Sans eliminates Linux font warnings)
plt.rcParams['font.sans-serif'] = 'DejaVu Sans'
fig, ax = plt.subplots(figsize=(8.8, 4.8), dpi=300)

labels = [
    f"{row['Category']}\n(n = {row['N']})" for _, row in summary.iterrows()
]
x = np.arange(len(summary))
width = 0.30

bars_like = ax.bar(
    x - width / 2,
    summary['Median_Like_Pct'],
    width,
    label='Shallow Engagement (Median Like Rate %)',
    color='#D1D5DB',
)
bars_intent = ax.bar(
    x + width / 2,
    summary['Median_Intent_Pct'],
    width,
    label='Actionable Intent (Median Save + Share + Comment Rate %)',
    color='#111827',
)

ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['left'].set_color('#9CA3AF')
ax.spines['bottom'].set_color('#9CA3AF')
ax.yaxis.grid(True, linestyle='--', linewidth=0.5, color='#E5E7EB', alpha=0.9)
ax.set_axisbelow(True)

plt.suptitle(
    'Utility-driven travel content generates 2.2x higher actionable intent'
    ' than high-ticket experiences',
    x=0.11,
    y=0.96,
    ha='left',
    fontsize=10.5,
    fontweight='bold',
    color='#111827',
)
ax.set_title(
    'Median engagement vs. actionable intent rates across TikTok China travel'
    f' videos (N = {len(df_clean)}; p < 0.001)',
    loc='left',
    fontsize=8.8,
    color='#4B5563',
    pad=12,
)

ax.set_ylabel('Median Rate (% of Views)', fontsize=9, color='#374151')
ax.set_xticks(x)
ax.set_xticklabels(labels, fontsize=8.8, color='#111827')
ax.set_ylim(0, 6.8)
ax.tick_params(axis='y', labelsize=8.5, colors='#4B5563')
ax.legend(frameon=False, loc='upper right', fontsize=8.5)

for bar in bars_like:
  h = bar.get_height()
  ax.annotate(
      f'{h:.2f}%',
      xy=(bar.get_x() + bar.get_width() / 2, h),
      xytext=(0, 3),
      textcoords='offset points',
      ha='center',
      fontsize=8,
      color='#6B7280',
  )

for bar in bars_intent:
  h = bar.get_height()
  ax.annotate(
      f'{h:.2f}%',
      xy=(bar.get_x() + bar.get_width() / 2, h),
      xytext=(0, 3),
      textcoords='offset points',
      ha='center',
      fontsize=8.2,
      fontweight='bold',
      color='#111827',
  )

fig.text(
    0.11,
    0.015,
    'Source: TikTok public video dataset scraped Nov 2026 (N = 400 deduplicated'
    ' videos across #chinatravel & 3 search queries).',
    fontsize=7.5,
    color='#6B7280',
)

plt.tight_layout(rect=[0.09, 0.04, 0.98, 0.91])
plt.savefig('Chili_Funnel_Benchmark.png', dpi=300)
plt.show()
```
